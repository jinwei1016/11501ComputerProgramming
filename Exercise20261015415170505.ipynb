習題一：使用random模組，產生出一個介於0到1之間的小數$r$，使用while迴圈，計算出$r$的前二十位二進位表示$r = 0.b_1b_2b_3\cdots {b_{20}}_2$。輸出格式如附圖Exercise20261015.png   
import random

r = random.random()
original_r = r

binary_digits = []
count = 0

while count < 20:
    r *= 2
    bit = int(r)
    binary_digits.append(str(bit))
    r -= bit
    count += 1

binary_str = "".join(binary_digits)
print(f"r = {original_r}")
print(f"r 的前二十位二進位表示: 0.{binary_str}")


Python 的 `int` 型別採用 Arbitrary-Precision Integer（任意精度整數）設計，能夠精確表示任意大小的整數，其實作方式可參考 CPython 的 `PyLongObject`。與 C/C++ 的 `int` 或 `long long` 不同，Python 的整數不受固定 32 位元或64 位元限制，而是會依據數值大小自動配置更多記憶體，因此只要記憶體足夠，便能表示極大的整數。例如：
```python
n = 10**68
print(n)
```
上述程式可正常執行，表示 Python 能夠精確儲存$10^{68}$，即中文數字中的「無量大數」。


可利用下列程式觀察不同大小整數所需的位元數（bit）與記憶體空間：

```python
import sys
n = 10**68
print(f"bit_length = {n.bit_length()}")
print(f"memory = {sys.getsizeof(n)} bytes")
```

並查閱官方文件，了解：

- `bit_length()`
- `sys.getsizeof()`
- `PyLongObject`

三者之間的關係。

習題二：輸入一個正整數 \(n\)，其中$1 \le n \le 10^{68}$，請將 \(n\) 轉換為中文大寫數字表示法。需支援下列數字單位：萬、億、兆、京、垓、秭、穰、溝、澗、正、載、極、恆河沙、阿僧祇、那由他、不可思議、無量大數。比如輸入31415926，輸出參仟壹佰肆拾壹萬伍仟玖佰貳拾陸。

提示：思考利用字典，建立「數字」與「中文大寫數字」的對應關係。
def number_to_chinese(n: int) -> str:
    if n == 0:
        return "零"
    
    digits = {0: '零', 1: '壹', 2: '貳', 3: '參', 4: '肆', 5: '伍', 6: '陸', 7: '柒', 8: '捌', 9: '玖'}
    units = ['', '拾', '佰', '仟']
    big_units = ['', '萬', '億', '兆', '京', '垓', '秭', '穰', '溝', '澗', '正', '載', '極', '恆河沙', '阿僧祇', '那由他', '不可思議', '無量大數']
    
    num_str = str(n)
    
    groups = []
    while num_str:
        groups.append(num_str[-4:])
        num_str = num_str[:-4]
    
    result_parts = []
    
    for idx, group in enumerate(groups):
        group_str = ""
        zero_flag = False
        g_len = len(group)
        
        for i, char in enumerate(group):
            digit = int(char)
            pos = g_len - 1 - i
            
            if digit != 0:
                if zero_flag:
                    group_str += digits[0]
                    zero_flag = False
                group_str += digits[digit] + units[pos]
            else:
                zero_flag = True
        
        if group_str:
            group_str += big_units[idx]
            result_parts.append(group_str)
        elif result_parts and zero_flag:
            result_parts.append(digits[0])
            
    final_result = "".join(reversed(result_parts))
    
    while "零零" in final_result:
        final_result = final_result.replace("零零", "零")
    final_result = final_result.rstrip("零")
    
    return final_result

num = 31415926
print(f"輸入：{num}")
print(f"輸出：{number_to_chinese(num)}")

outer_iter = iter([1, 2, 3, 4, 5, 6, 7, 8, 9])

while True:
